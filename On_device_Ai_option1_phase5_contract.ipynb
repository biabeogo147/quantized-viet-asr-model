# BKMeeting AI Hub Option 1 Phase 5 Contract Packaging

This notebook is the dedicated `Phase 5` packaging notebook for `Option 1`.

Use it after Phase 4 records already exist.
It should never compile models and it should not rerun Phase 4 when the gate records already exist.


## Environment Notes

Before running this notebook, make sure the current environment can already execute the local `python-model-test` helpers and that the target records for the chosen `RUN_LABEL` already exist.

This notebook is package-only.


```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd() / "src"))

from tools.aihub_option1_phase5_contract import materialize_phase5_contract_package
from tools.aihub_option1_pilots import build_option1_runtime_config
```

## Phase 5 Config


```python
DEVICE_NAME = "Samsung Galaxy S24 (Family)"
QAIRT_VERSION = None
RUN_LABEL = "option1-all-20260513"

ENABLE_ZIPFORMER = True
ENABLE_VPCD = True

PHASE5_PACKAGE_LABEL = None
PHASE5_OUTPUT_ROOT = None

RUNTIME_CONFIG = build_option1_runtime_config(
    device_name=DEVICE_NAME,
    qairt_version=QAIRT_VERSION,
    repo_root=Path.cwd(),
)

print("record_root:", RUNTIME_CONFIG.record_root)
print("run_label:", RUN_LABEL)
print("enable zipformer:", ENABLE_ZIPFORMER)
print("enable vpcd:", ENABLE_VPCD)
print("phase5 package label:", PHASE5_PACKAGE_LABEL)
print("phase5 output root override:", PHASE5_OUTPUT_ROOT)
```

### Package Zipformer Phase 5 Contract


```python
zipformer_phase5_package_report = None
if PHASE5_INCLUDE_ZIPFORMER:
    zipformer_phase5_package_report = materialize_phase5_contract_package(
        pilot_name='zipformer',
        runtime_config=RUNTIME_CONFIG,
        run_label=RUN_LABEL,
        package_label=PHASE5_PACKAGE_LABEL,
        output_root=PHASE5_OUTPUT_ROOT,
    )
    print('zipformer phase5 package:', zipformer_phase5_package_report['package_path'])
    print('zipformer phase5 promotion status:', zipformer_phase5_package_report['promotion_status'])
    print('zipformer phase5 warnings:', zipformer_phase5_package_report['warnings'])
else:
    print('Skipping Zipformer Phase 5 packaging because PHASE5_INCLUDE_ZIPFORMER is False.')
```

### Package VPCD Phase 5 Contract


```python
vpcd_phase5_package_report = None
if PHASE5_INCLUDE_VPCD:
    vpcd_phase5_package_report = materialize_phase5_contract_package(
        pilot_name='vpcd',
        runtime_config=RUNTIME_CONFIG,
        run_label=RUN_LABEL,
        package_label=PHASE5_PACKAGE_LABEL,
        output_root=PHASE5_OUTPUT_ROOT,
    )
    print('vpcd phase5 package:', vpcd_phase5_package_report['package_path'])
    print('vpcd phase5 promotion status:', vpcd_phase5_package_report['promotion_status'])
    print('vpcd phase5 warnings:', vpcd_phase5_package_report['warnings'])
else:
    print('Skipping VPCD Phase 5 packaging because PHASE5_INCLUDE_VPCD is False.')
```

## Phase 5 Packaging Summary


```python
print('phase5 packaging summary:')
for label, report in [
    ('zipformer', globals().get('zipformer_phase5_package_report')),
    ('vpcd', globals().get('vpcd_phase5_package_report')),
]:
    if report is None:
        print(label, 'not packaged in this run.')
        continue
    print(
        {
            'pilot': label,
            'package_path': str(report['package_path']),
            'promotion_status': report['promotion_status'],
            'warnings': report['warnings'],
        }
    )

print('phase5 reminder: package creation does not override the Phase 4 verdict.')
```
